## 1.1 Kommentarer

Når vi skriver kode ønsker vi ofte å forklare hva koden gjør, både for oss selv og andre som skal lese den senere. Da bruker vi **kommentarer**. Kommentarer blir ikke kjørt som kode av Python.

Det finnes to måter å skrive kommentarer på i Python:

**1. Med hash (`#`)**  
Alt som står etter `#` på samme linje blir ignorert av Python.

**2. Med trippel anførselstegn (`"""`)**  
Brukes når du vil kommentere over flere linjer samtidig.

```python
# Dette er en kommentar - Python leser ikke denne linjen som kode
print("Hei verden")  # Dette er også en kommentar, etter koden på samme linje

"""
Dette er en kommentar
som strekker seg
over flere linjer
"""
```

## 1.2 print()-funksjonen

`print()` er en av de mest brukte funksjonene i Python. Den skriver ut det du ber om i konsollen (output-feltet under kodecellen).

`print()` kan ta imot ett eller flere argumenter (verdier). Dersom du vil skrive ut flere ting på samme linje, skiller du dem med komma.

## 1.3 Datatyper

I Python finnes det flere datatyper. De mest brukte er:

| Datatype | Beskrivelse | Eksempel |
|---|---|---|
| **int** (heltall) | Hele tall, uten desimaler | `5`, `-3`, `200` |
| **float** (flyttall) | Tall med desimaler | `3.14`, `-0.5` |
| **str** (streng/tekst) | Tekst, skrives i anførselstegn | `"Hei"`, `"Python"` |
| **bool** (boolsk verdi) | Sann/usann | `True`, `False` |
| **list** (liste) | Samling av flere verdier | `[1, 2, 3]` |

Vi kommer tilbake til lister i en senere notebook.

## 1.4 Variabler

En **variabel** er et navn vi gir til en verdi, slik at vi kan bruke verdien senere i programmet. Vi oppretter en variabel med likhetstegn (`=`).

`variabelnavn = verdi`

Det som står til høyre for likhetstegnet blir lagt inn i variabelen på venstre side.

```python
a = 4                 # a er nå et heltall (int)
print(a)

b = 2.718             # b er nå et flyttall (float)
print(b)

c = "Jeg er en tekst" # c er nå en tekststreng (str)
print(c)

d = True              # d er en boolsk verdi (bool)
print(d)
```

Vi kan sjekke hvilken datatype en variabel har med funksjonen `type()`.

```python
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

Vi kan også gjøre om (konvertere) mellom datatyper med funksjonene `int()`, `float()` og `str()`.

| Funksjon | Gjør om til | Eksempel |
|---|---|---|
| `int()` | heltall | `int(2.718)` gir `2` |
| `float()` | flyttall | `float(3)` gir `3.0` |
| `str()` | tekst | `str(34)` gir `"34"` |

```python
e = int(2.718)     # gjør om til heltall (kutter desimalene)
print(e)

f = float(3)       # gjør om til flyttall
print(f)

g = str(34)        # gjør om til tekst
print(g)
print(type(g))
```

## 1.5 Aritmetiske operasjoner

Python kan brukes som en avansert kalkulator. Her er de vanligste regneoperatorene:

| Operator | Betydning | Eksempel |
|---|---|---|
| `+` | Addisjon | `4 + 2` → `6` |
| `-` | Subtraksjon | `4 - 2` → `2` |
| `*` | Multiplikasjon | `4 * 2` → `8` |
| `/` | Divisjon | `4 / 2` → `2.0` |
| `**` | Eksponent (opphøyd i) | `4 ** 2` → `16` |
| `//` | Heltallsdivisjon | `7 // 2` → `3` |
| `%` | Modulo (rest ved divisjon) | `7 % 2` → `1` |

```python
a = 4
b = 2.718

print(a + b)   # Addisjon
print(a - b)   # Subtraksjon
print(a * b)   # Multiplikasjon
print(a / b)   # Divisjon
print(a ** 2)  # Eksponent
print(a % b)   # Modulus (rest ved divisjon)
```

**Heltallsdivisjon (`//`)** forteller oss hvor mange hele ganger et tall går opp i et annet.  
**Modulo (`%`)** forteller oss hva som er *igjen* (resten) etter en slik divisjon.

```python
print(7 // 2)   # Hvor mange hele ganger går 2 opp i 7? Svar: 3
print(7 % 2)    # Hva blir resten? Svar: 1
```
